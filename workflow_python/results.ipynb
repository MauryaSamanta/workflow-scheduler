```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the CSV
df = pd.read_csv('deadline_vs_cost.csv')

# Optional: sort the deadline_factor just to make sure plot line is smooth
df = df.sort_values(by='deadline_multiplier')

# Plot setup
plt.figure(figsize=(10, 6))
sns.set_style('whitegrid')

# Plot each workflow as separate line
for workflow_name in df['workflow'].unique():
    sub_df = df[df['workflow'] == workflow_name]
    plt.plot(sub_df['deadline_multiplier'], sub_df['total_cost'], marker='o', label=workflow_name)

plt.xlabel("Deadline Factor (x)")
plt.ylabel("Total Cost ($)")
plt.title("Cost vs Deadline Factor for Workflows")
plt.legend(title="Workflow")
plt.tight_layout()
plt.show()
```
